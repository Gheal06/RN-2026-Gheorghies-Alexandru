## Benchmark different ways of calculating the size of a tree

```python
import random
from tqdm.auto import trange
```

Setup for the experiment. You may play with these variable to set up different experiments.

```python
RANDOM_STATE = 42

N_TREES = 100
MIN_CHILDREN = 0
MAX_CHILDREN = 10
MAX_DEPTH = 12

LEAF_PROBABILITY = 0.40

NUMBER = 100
REPEAT = 5
```

The `Tree` class.

```python

import itertools
import functools
import sys

class Tree:
    def __init__(self, subrules=None) -> None:
        self.subrules = [] if subrules is None else subrules

    def fastest_len_wa(self) -> int:
        q = [self] 
        
        return functools.reduce(lambda acc, val: (
            sys._getframe(1).f_locals.__setitem__('q',itertools.chain(q, val.subrules)), acc+1
        )[-1],
        q, 
        0)
    
    def fastest_len(self) -> int:
        q = [[self]]
        return functools.reduce(lambda acc, val: (
            q.append(val.subrules),acc+1
        )[-1],
        itertools.chain.from_iterable(q), 
        0)
    
    def second_fastest_len(self) -> int:
        q = [self]
        for u in q:
            q.extend(u.subrules)
        return len(q)

    def len_dfs_recursive(self) -> int:
        return 1 + sum(subrule.len_dfs_recursive() for subrule in self.subrules)

print(Tree([Tree(),Tree()]).fastest_len_wa())
```

```python
def create_random_tree(
    rng: random.Random,
    depth: int = 0,
) -> Tree:
    if depth >= MAX_DEPTH or rng.random() < LEAF_PROBABILITY:
        return Tree()

    n_children = min(
        rng.randint(MIN_CHILDREN, MAX_CHILDREN), rng.randint(MIN_CHILDREN, MAX_CHILDREN)
    )

    return Tree([create_random_tree(rng, depth + 1) for _ in range(n_children)])

rng = random.Random(RANDOM_STATE)
trees = [create_random_tree(rng) for _ in trange(N_TREES, desc="creating trees")]
```
